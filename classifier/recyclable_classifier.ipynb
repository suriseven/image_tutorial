```python
from ultralytics import YOLO

model = YOLO("yolo11n-cls.pt")

# recycle/재활용 쓰레기 이미지 데이터 안에 있는 각 디렉토리를 라벨로 하여 이미지들을 학습합니다
# epochs: 반복 학습 횟수
# imgsz
#  - 학습할때 리사이즈해서 사용할 이미지 크기
#  - 제공되는 데이터의 가로/세로 통합 최대 크기를 기준으로 320으로 잡았습니다
#  - 만약 직접 촬영하는 이미지의 크기가 이보다 더 크다면 640정도로 늘려보는것도 좋습니다
model.train(
    data="recycle/재활용 쓰레기 이미지 데이터",
    epochs=10,
    imgsz=320
)
```

```python
import os

# 이 데이터는 학습에 사용하지 않습니다
# 촬영하신 사진을 테스트 데이터로 사용하기 원하신다면 이 디렉토리 안에 복사해주세요
# 학습된 모델을 이용해 디렉토리 안에 있는 모든 이미지를 평가하고 시각화 합니다
test_dir =  'recycle/테스트 데이터'

test_files = []

for root, dirs, files in os.walk(test_dir):
    for f in files:
        file_path = os.path.join(root, f)
        test_files.append(file_path)

results = model(test_files)

# 시각화 결과의 의미는 각 분류 결과에 대한 확률 입니다
# 분류 모델은 여기서 가장 높은 확률의 결과로 분류합니다
# 예를 들어 아래와 같은 경우 캔으로 분류되었다고 보면 됩니다
# 캔 0.39
# 비닐 0.27
# 유리 0.10
# 플라스틱 0.10
# 종이 0.09
for result in results:
    result.show()
```
